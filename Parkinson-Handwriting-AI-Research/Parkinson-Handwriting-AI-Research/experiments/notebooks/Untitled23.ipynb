```python
spiral_dir = "/content/drive/MyDrive/Spiral_HandPD"
meander_dir = "/content/drive/MyDrive/Meander_HandPD"

IMG_SIZE = 224
```

```python
data = []
labels = []

def load_images(folder, label):

    for img_name in os.listdir(folder):

        if img_name.lower().endswith(('.png','.jpg','.jpeg')):

            img_path = os.path.join(folder,img_name)

            img = cv2.imread(img_path)

            if img is None:
                continue

            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img = cv2.resize(img,(IMG_SIZE,IMG_SIZE))

            data.append(img)
            labels.append(label)
```

```python
load_images(os.path.join(spiral_dir,"healthy"),0)
load_images(os.path.join(spiral_dir,"parkinson"),1)

load_images(os.path.join(meander_dir,"healthy"),0)
load_images(os.path.join(meander_dir,"parkinson"),1)

X = np.array(data)
y = np.array(labels)

print("Before pHash:",X.shape)
```
